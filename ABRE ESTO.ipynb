La "Pesca por Voz" es un minijuego interactivo en el que el jugador controla una caña de pescar mediante su voz. En lugar de utilizar teclado o ratón, la posición y profundidad del anzuelo se ajustan en tiempo real a partir del tono y volumen del sonido que emite el jugador.

Este enfoque propone una experiencia inmersiva y lúdica que convierte la voz en una herramienta de control, permitiendo explorar los conceptos del procesamiento de señales de audio en una situación práctica. A través de la detección de tonos musicales y la medición del volumen en decibelios, el juego ofrece una manera intuitiva de "cantar" para pescar.

<br>

<div style="color:#7d8ea4;; padding:10px; border-radius:10px">
  <b>Análisis espectral mediante FFT</b>
</div>
El juego emplea una <b>Transformada Rápida de Fourier (FFT)</b> para analizar el audio capturado en tiempo real y detectar la frecuencia fundamental del sonido emitido por el jugador. Este proceso es clave en el análisis de señales digitales, permitiendo estimar el tono y asociarlo a una nota musical. 

<div style="color:#7d8ea4;; padding:10px; border-radius:10px">
  <b>Detección de notas musicales</b>
</div>
A partir del tono detectado, se identifica la nota musical más cercana utilizando un mapa de frecuencias. Esta asignación permite que el jugador controle la posición horizontal del anzuelo, lo que traduce el análisis espectral en una acción visual directa.  

<div style="color:#7d8ea4;; padding:10px; border-radius:10px">
  <b>Procesamiento de volumen (RMS y dB)</b>
</div>
El volumen de la voz se mide utilizando la Raíz Cuadrática Media (RMS) y se convierte a decibelios (dB), una práctica habitual en análisis físico y perceptivo de señales. El volumen determina la longitud del sedal, lo que representa la profundidad del anzuelo en el agua.

<div style="color:#7d8ea4;; padding:10px; border-radius:10px">
  <b>Control por voz e interacción en tiempo real</b>
</div>
Este proyecto demuestra cómo la voz puede ser utilizada como una interfaz de control natural. El tono guía la dirección del anzuelo (izquierda-derecha), y el volumen define su profundidad (más fuerte, más abajo), generando una interacción continua y significativa entre el jugador y el sistema.

En este juego, controlas una caña de pescar únicamente con tu voz. No necesitas teclado ni ratón: el micrófono interpreta tanto el tono como el volumen del sonido que emites.

El **tono** (qué tan agudo o grave cantas) determina la posición **horizontal** del anzuelo. Cantar más agudo lo mueve a la derecha, y más grave lo desplaza hacia la izquierda. Solo se consideran los tonos dentro de una octava activa, que puedes cambiar durante el juego.

El **volumen** de tu voz controla la **profundidad** del sedal: cuanto más fuerte hables o cantes, más baja el anzuelo; si guardas silencio, la caña se recoge. Esta relación entre sonido e interacción convierte tu voz en un control sensible y dinámico.

El objetivo es atrapar peces, que nadan por la pantalla. Al capturar varios, aparece un **pez especial** (se trata de una tortuga) que vale más puntos. Una interfaz visual te ayuda mostrando el volumen, la nota detectada y la octava activa. Además, puedes pausar con P, ajustar la velocidad de los peces con los botones + y –, o cambiar la octava con las flechas.

Este juego es una forma divertida de explorar cómo el procesamiento de voz puede integrarse en la interacción con sistemas multimedia. Solo necesitas un micrófono, tu voz y ganas de cantar.

<div style="color:#b38766; padding:10px; border-radius:10px">
  <b>Procesamiento digital de señales de voz</b>
</div>

El juego analiza continuamente el volumen de la señal captada por el micrófono. Para ello, se calcula la energía de la señal, uno de los primeros pasos habituales en cualquier sistema de codificación, reconocimiento o análisis de voz. Esta extracción de características es una aplicación directa de los conceptos estudiados.

<div style="color:#b38766; padding:10px; border-radius:10px">
  <b>Psicoacústica</b>
</div>

El nivel de sonido no se calcula de forma directa, sino usando una media ponderada, teniendo en cuenta que el oído humano no percibe igual un sonido breve que uno sostenido. Este enfoque permite una interacción más suave y natural, evitando respuestas indeseadas ante ruidos puntuales y reduciendo la fatiga auditiva del jugador.

<div style="color:#b38766; padding:10px; border-radius:10px">
  <b>Muestreo y codificación</b>
</div>

Para capturar la señal de voz se utiliza una frecuencia de muestreo de **44.1 kHz**, igual a la empleada en estándares profesionales como CD-Audio o archivos WAV. Esta elección no solo garantiza una calidad suficiente, sino que conecta directamente con los temas de codificación y fidelidad en la digitalización del sonido.

<div style="color:#b38766; padding:10px; border-radius:10px">
  <b>Interacción y control por voz</b>
</div>

El proyecto explora la voz como interfaz de control, una tendencia en auge en ámbitos como la accesibilidad, los videojuegos sin manos (hands-free) o la realidad aumentada. El uso del volumen vocal como entrada demuestra cómo el audio puede convertirse en una forma de interacción intuitiva y dinámica.

<div style="color:#6ea185; padding:10px; border-radius:10px">
  <b>Análisis espectral mediante FFT</b>
</div>

El juego detecta el tono de la voz aplicando una Transformada Rápida de Fourier (FFT) con una ventana de Hanning, una técnica esencial en el análisis espectral de señales. Esto permite extraer la frecuencia fundamental de lo que el jugador canta, y asociarla a una nota musical concreta.

<div style="color:#6ea185; padding:10px; border-radius:10px">
  <b>Filtrado y rango vocal útil</b>
</div>

Para mejorar la precisión, se ignoran las frecuencias fuera del rango vocal humano útil (200–700 Hz). Esta selección actúa como un filtro de paso banda y se basa en principios de procesamiento de señales de voz, ayudando a reducir el ruido y evitar detecciones incorrectas.

<div style="color:#6ea185; padding:10px; border-radius:10px">
  <b>Reconocimiento de notas y percepción sonora</b>
</div>

El sistema utiliza umbrales de volumen y una tolerancia de frecuencia para reconocer las notas. Esto se relaciona directamente con la percepción humana del sonido, como el umbral de detección y la capacidad para distinguir entre tonos cercanos, especialmente en frecuencias más altas, donde la tolerancia se amplía para adaptarse a las limitaciones perceptuales.

<div style="color:#6ea185; padding:10px; border-radius:10px">
  <b>Síntesis de audio</b>
</div>

Los sonidos que emite el juego para representar cada nota se generan digitalmente como ondas sinusoidales, aplicando un principio básico de síntesis de audio. Aunque no se emplean estándares de codificación como MPEG o AAC, se trabaja con muestras de audio digitalizadas a 44100 Hz, 16 bits y en mono, entendiendo así la estructura y almacenamiento de señales sonoras.

Simon Dice Vocal es un juego en el que tienes que repetir, con tu voz, una secuencia de notas musicales que el sistema reproduce con sonido y color. Cada vez que aciertas la secuencia completa, se añade una nota más, aumentando la dificultad progresivamente.

El juego utiliza estos cuatro colores y notas:

🔴 Do4 (261.63 Hz)

🟢 Fa4 (349.23 Hz)

🔵 La4 (440 Hz)

🟡 Do5 (523.25 Hz)

¿Cómo se juega?

- **Observa y escucha**: El juego te muestra la secuencia actual de notas, iluminando colores y reproduciendo los sonidos correspondientes.

- **Canta en orden**: Tienes que repetir esa secuencia cantando las notas en el mismo orden. El micrófono capta tu voz en tiempo real.

- **Análisis de voz**: El sistema aplica una FFT para detectar la frecuencia fundamental y la compara con la nota esperada. Si está dentro del margen de tolerancia, se considera correcta.

- **Sigue avanzando**: Si aciertas toda la secuencia, pasas al siguiente nivel. Si fallas, el juego muestra el mensaje “¡Fallaste! Reiniciando...” y vuelve a empezar desde la primera nota.

Para que una nota se registre correctamente, debes mantenerla brevemente. El juego también registra tu **récord personal**: la mayor cantidad de notas seguidas que lograste repetir sin fallar.

Además, puedes pedir una **pista**: si sabes qué nota debes cantar pero quieres volver a escucharla para afinar mejor, puedes hacer clic sobre el cuadrado correspondiente a esa nota. El sistema reproducirá nuevamente su sonido, ayudándote a recordarla.
Pero **CUIDADO**: si haces clic en un cuadrado que no corresponde con la nota esperada en la secuencia actual, el juego lo considerará un error, y perderás la partida.

<div style="color:#b45665; padding:10px; border-radius:10px">
  <b>Digitalización y Procesamiento de Audio</b>
</div>

**Muestreo y Representación Digital**:
El juego captura la voz en tiempo real a través del micrófono, convirtiendo una señal acústica (analógica) en datos digitales. Para ello se utiliza una frecuencia de muestreo estándar de 44100 Hz, trabajando directamente con muestras PCM. Esta digitalización es uno de los pilares de la asignatura, ya que representa la conversión de señales del mundo real a un formato que el ordenador puede interpretar.

**Parámetros de Ondas Acústicas**:
La voz del jugador se analiza a partir de su amplitud, es decir, el volumen de la señal. Para ello se calcula la Raíz Cuadrática Media (RMS), que luego se transforma en decibelios (dB). Estos conceptos permiten cuantificar la intensidad sonora y son fundamentales para el análisis físico de las señales.


<div style="color:#b45665; padding:10px; border-radius:10px">
  <b>Procesamiento Básico de la Voz para la Interacción</b>
</div>

Análisis de Amplitud en el Tiempo:
En lugar de analizar el contenido en frecuencias, el juego emplea un enfoque más simple y eficiente: el análisis de la amplitud a lo largo del tiempo.
Dependiendo de cómo varíe esta amplitud, se interpretan dos comandos distintos:

    Disparo: Detectado cuando hay un pico breve de volumen (un sonido corto e intenso).

    Barrera: Activada con un nivel de volumen alto y sostenido durante un tiempo determinado.

Este tipo de análisis permite reconocer comandos vocales sin necesidad de recurrir a algoritmos complejos, aplicando principios básicos del procesado de voz.

El objetivo es proteger la nave de los enemigos utilizando la voz para disparar y activar una barrera, mientras recolectas monedas y subes de nivel.

Presiona **ESPACIO** en la pantalla de inicio para comenzar.

Mueve la nave con las **flechas** izquierda y derecha para esquivar enemigos y recoger monedas.

Para disparar, emite un **sonido corto** e intenso (preferiblemente un golpe o palmada). El disparo tiene un cooldown (tiempo de espera) y no está disponible mientras la barrera está activa.

Para activar la **barrera**, emite un **sonido fuerte** y sostenido. También tiene cooldown. La barrera protege de enemigos pero desactiva temporalmente los disparos.

Los enemigos bajan desde la parte superior. Si colisionan con la nave sin barrera activa, el juego termina.Sin embargo, si en ese momento has acumulado al menos 10 monedas, recibirás automáticamente una vida extra y podrás continuar la partida una vez más a cambio de esas 10 oros.

Avanzas de nivel cada vez que eliminas **5 enemigos**, hasta un máximo de nivel 10. Cada nivel incrementa la dificultad.

Puedes recoger monedas para mejorar tu puntuación final.

La interfaz (HUD) muestra en todo momento:

- Nivel actual

- Número de eliminaciones

- Monedas recolectadas

- Estado del disparo y la barrera (activo/inactivo)

- Barra de volumen en tiempo real

Cuando pierdes, aparece una pantalla de Game Over con estadísticas finales y un gráfico que muestra el historial de volumen y los momentos de disparo y barrera.
Presiona **'R'** para reiniciar la partida.
